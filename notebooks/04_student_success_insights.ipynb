# Student Success Analysis

The purpose of this notebook is to translate machine learning findings into actionable educational insights.

The analysis focuses on:

- Characteristics of high-performing students
- Characteristics of low-performing students
- Key drivers of academic success
- Potential intervention strategies

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("../data/student_dataset_10000_rows.csv")
```

## Performance Categories

Students were divided into three performance groups based on their final exam scores.

- Needs Support: Score below 70
- Average: Score between 70 and 85
- High Performer: Score above 85

These groups allow us to compare academic behaviors across different performance levels.

```python
df["performance_group"] = pd.cut(
    df["exam_score"],
    bins=[0, 70, 85, 100],
    labels=[
        "Needs Support",
        "Average",
        "High Performer"
    ],
    include_lowest=True
)
```

```python
df["performance_group"].value_counts()
```

```python
sns.countplot(
    data=df,
    x="performance_group"
)
plt.title("Distribution of Performance Groups")
plt.show()
```

### Observation

The majority of students fall into the High Performer category.

This finding is consistent with the score distribution observed during exploratory analysis, where many students achieved scores near 100.

```python
sns.boxplot(
    data=df,
    x="performance_group",
    y="study_hours"
)
```

### Observation

High-performing students generally study more hours than average students and students requiring support.

Study hours appear to be one of the strongest differentiators between performance groups.

```python
sns.boxplot(
    data=df,
    x="performance_group",
    y="attendance"
)
```

Students in the High Performer group generally maintain higher attendance levels compared to lower-performing groups.

```python
sns.boxplot(
    data=df,
    x="performance_group",
    y="assignments_completed"
)
```

Assignment completion rates increase consistently as academic performance improves.

```python
sns.boxplot(
    data=df,
    x="performance_group",
    y="internet_usage"
)
```

Students requiring support tend to spend more time using the internet, although the relationship is weaker than study hours or assignment completion.

```python
summary = df.groupby(
    "performance_group",
    observed=False
).agg({
    ...
})

summary.round(2)
```

```python
plt.figure(figsize=(10,6))

sns.heatmap(
    summary,
    annot=True,
    cmap="Blues"
)

plt.title(
    "Average Characteristics by Performance Group"
)

plt.show()
```

# Key Findings

## Academic Success Drivers

The analysis revealed several characteristics consistently associated with stronger academic performance:

1. Increased study hours
2. Higher assignment completion rates
3. Better historical academic performance
4. Stronger attendance records

## Characteristics of High-Performing Students

Compared to students requiring support, high-performing students:

- Study more hours
- Complete more assignments
- Have stronger previous academic scores
- Maintain higher attendance levels

## Potential Educational Interventions

Based on the findings, educational institutions may improve student outcomes by:

- Encouraging consistent study habits
- Increasing assignment engagement
- Monitoring attendance more closely
- Providing early support to students with weaker historical performance

## Conclusion

The factors identified in this analysis align closely with the variables found to be most important during machine learning modeling, reinforcing the importance of academic engagement as a predictor of student success.
